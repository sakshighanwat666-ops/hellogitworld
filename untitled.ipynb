```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    'Name': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Age': [20, 21, 22, 23, 21, 24],
    'Marks': [75, 85, 65, 90, 80, 70],
    'Hours': [2, 4, 3, 5, 4, 2]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df['Name'], df['Marks'], color='skyblue')

plt.title('Student Marks')
plt.xlabel('Students')
plt.ylabel('Marks')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Hours'], df['Marks'], color='green')

plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Marks'], bins=5, color='orange', edgecolor='black')

plt.title('Distribution of Marks')
plt.xlabel('Marks')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(y=df['Marks'])

plt.title('Box Plot of Student Marks')

plt.show()
```

```python
plt.figure(figsize=(6, 4))

correlation = df[['Age', 'Marks', 'Hours']].corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].bar(df['Name'], df['Marks'], color='skyblue')
axes[0, 0].set_title('Bar Chart')

axes[0, 1].scatter(df['Hours'], df['Marks'])
axes[0, 1].set_title('Scatter Plot')

axes[1, 0].hist(df['Marks'], bins=5, edgecolor='black')
axes[1, 0].set_title('Histogram')

sns.boxplot(y=df['Marks'], ax=axes[1, 1])
axes[1, 1].set_title('Box Plot')

plt.tight_layout()
plt.show()
```
